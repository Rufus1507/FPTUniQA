# 04. Tìm Kiếm Lai (Hybrid Retrieval) & RRF & Re-ranking

Notebook này phân tích sự kết hợp giữa Sparse (BM25) và Dense (FAISS) qua thuật toán Reciprocal Rank Fusion và Cross-Encoder Reranker.

```python
from university_qa.retrieval.retriever import HybridRetriever
from university_qa.reranking.cross_encoder import CrossEncoderReranker
from university_qa.utils.io import read_jsonl

docs = read_jsonl('../data/processed/corpus.jsonl')
hybrid = HybridRetriever()
hybrid.index(docs)

query = 'đăng ký bao nhiêu tín chỉ một học kỳ'
hybrid_results = hybrid.retrieve(query, top_k=5)
candidates = [d for d, _ in hybrid_results]

reranker = CrossEncoderReranker()
final = reranker.rerank(query, candidates, top_n=3)
for doc, score in final:
    print(f'Rerank Score: {score:.4f} - {doc["title"]}')
```
