# 03. Tìm Kiếm Ngữ Nghĩa Dense Vector Với FAISS

Notebook này phân tích vector embedding và tìm kiếm lân cận gần nhất (kNN) qua FAISS.

```python
from university_qa.retrieval.dense import DenseRetriever
from university_qa.utils.io import read_jsonl

docs = read_jsonl('../data/processed/corpus.jsonl')
dense = DenseRetriever()
dense.index(docs)

results = dense.retrieve('điểm F xử lý thế nào', top_k=3)
for doc, score in results:
    print(f'Score: {score:.4f} - {doc["title"]}')
```
