# 05. Đo Lường & Trực Quan Hóa Kết Quả Thực Nghiệm (Evaluation)

Notebook này biểu diễn đồ thị so sánh giữa BM25 Baseline, Dense Retrieval và Hybrid + Reranker.

```python
import matplotlib.pyplot as plt
import json

# Dữ liệu minh họa benchmark đối sánh
models = ['BM25 Baseline', 'Dense (FAISS)', 'Hybrid (RRF)', 'Hybrid + Reranker']
mrr_scores = [0.62, 0.74, 0.81, 0.89]
hit_at_3 = [0.70, 0.82, 0.88, 0.95]

fig, ax = plt.subplots(figsize=(8, 4))
x = range(len(models))
ax.bar([i - 0.2 for i in x], mrr_scores, width=0.4, label='MRR@20', color='#3B82F6')
ax.bar([i + 0.2 for i in x], hit_at_3, width=0.4, label='Hit@3', color='#10B981')
ax.set_xticks(list(x))
ax.set_xticklabels(models)
ax.set_ylim(0, 1.0)
ax.set_title('So Sánh Hiệu Năng Các Kiến Trúc Retrieval')
ax.legend()
plt.show()
```
