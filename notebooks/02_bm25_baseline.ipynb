# 02. Xây Dựng & Kiểm Thử Mô Hình Cơ Sở BM25

Notebook này đánh giá hiệu năng của bộ tìm kiếm từ khóa BM25 Okapi trên tập văn bản tiếng Việt.

```python
from university_qa.retrieval.bm25 import BM25Retriever
from university_qa.utils.io import read_jsonl

docs = read_jsonl('../data/processed/corpus.jsonl')
bm25 = BM25Retriever()
bm25.index(docs)

query = 'Sinh viên bị cảnh báo học tập mấy lần thì bị đuổi học?'
results = bm25.retrieve(query, top_k=3)
for doc, score in results:
    print(f'[{score:.2f}] {doc["title"]}')
```
