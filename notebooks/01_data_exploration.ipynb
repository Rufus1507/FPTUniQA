Notebook này phân tích cấu trúc tài liệu quy chế đại học, phân bố độ dài văn bản và số lượng chunk.

```python
from university_qa.utils.io import read_jsonl
import pandas as pd

# Đọc corpus dữ liệu
corpus = read_jsonl('../data/processed/corpus.jsonl')
df = pd.DataFrame(corpus)
print(f'Tổng số documents/chunks: {len(df)}')
df.head()
```

```python
# Phân bố độ dài ký tự
df['char_length'] = df['text'].apply(len)
print(df['char_length'].describe())
```
